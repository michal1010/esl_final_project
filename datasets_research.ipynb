### Diabetes 130 US hospitals for years 1999-2008

**good dataset**

```python
path = kagglehub.dataset_download("brandao/diabetes")

print("Path to dataset files:", path)
```

```python
dataset = pd.read_csv(Path(path) / "diabetic_data.csv")
```

```python
dataset.head()
```

```python
dataset
```

```python
dataset = dataset.dropna(subset=['race'])
```

```python
dataset['readmitted'] = dataset['readmitted'].apply(lambda x: 'NO' if x == 'NO' else 'YES')
```

```python
dataset.groupby('race')['readmitted'].value_counts()
```

```python
for race in dataset['race'].unique():
    filtered_yes = dataset[(dataset['race'] == race) & (dataset['readmitted'] == 'YES')]
    filtered_no = dataset[(dataset['race'] == race) & (dataset['readmitted'] == 'NO')]

    print(f"Race: {race}, Readmitted YES/NO count: {len(filtered_yes)/len(filtered_no)}")
```

```python
columns_to_encode = ['race', 'insulin']
```

```python
data_dummies = dataset[['race', 'insulin']]
```

```python
data_dummies = pd.get_dummies(data_dummies, columns=['race', 'insulin'], drop_first=False)
data_dummies
```

```python

covariance_matrix = data_dummies.corr()
covariance_matrix.style.background_gradient(cmap='coolwarm', axis=None, vmin=-1, vmax=1).format(precision=3)
```

```python
data_dummies = pd.get_dummies(data_dummies, columns=['sex', 'race', 'c_charge_degree', 'c_charge_desc'], drop_first=False)
data_dummies
```
