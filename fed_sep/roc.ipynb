```python
plt.plot(
    df.loc[df['race'] == 'Hispanic']['FPR'].values,
    df.loc[df['race'] == 'Hispanic']['TPR'].values,
)

plt.plot(
    df.loc[df['race'] == 'Asian']['FPR'].values,
    df.loc[df['race'] == 'Asian']['TPR'].values,
)

plt.plot(
    df.loc[df['race'] == 'Caucasian']['FPR'].values,
    df.loc[df['race'] == 'Caucasian']['TPR'].values,
)

plt.plot(
    df.loc[df['race'] == 'Other']['FPR'].values,
    df.loc[df['race'] == 'Other']['TPR'].values,
)

plt.plot(
    df.loc[df['race'] == 'African-American']['FPR'].values,
    df.loc[df['race'] == 'African-American']['TPR'].values,
)

plt.plot(
    df.loc[df['race'] == 'Native American']['FPR'].values,
    df.loc[df['race'] == 'Native American']['TPR'].values,
)

sep_points = pd.read_csv('metrics.csv')

plt.scatter(
    sep_points.loc[sep_points['Race'] == 'Hispanic']['FPR'].values,
    sep_points.loc[sep_points['Race'] == 'Hispanic']['TPR'].values,
    marker="x"
)

plt.scatter(
    sep_points.loc[sep_points['Race'] == 'Asian']['FPR'].values,
    sep_points.loc[sep_points['Race'] == 'Asian']['TPR'].values,
    marker="x"
)

plt.scatter(
    sep_points.loc[sep_points['Race'] == 'Caucasian']['FPR'].values,
    sep_points.loc[sep_points['Race'] == 'Caucasian']['TPR'].values,
    marker="x"
)

plt.scatter(
    sep_points.loc[sep_points['Race'] == 'Other']['FPR'].values,
    sep_points.loc[sep_points['Race'] == 'Other']['TPR'].values,
    marker="x"
)

plt.scatter(
    sep_points.loc[sep_points['Race'] == 'African-American']['FPR'].values,
    sep_points.loc[sep_points['Race'] == 'African-American']['TPR'].values,
    marker="x"
)

plt.scatter(
    sep_points.loc[sep_points['Race'] == 'Native American']['FPR'].values,
    sep_points.loc[sep_points['Race'] == 'Native American']['TPR'].values,
    marker="x"
)

plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('Federated model ROC with separation points.')
plt.legend(['Hispanic', 'Asian', 'Caucasian', 'Other', 'African-American', 'Native American'])


```
